## **Dataset preprocessing**

```python

#Listing the data path
real_videos = list(Path("/content/drive/MyDrive/data/face++dataset/ffpp_real").glob("*.mp4"))
fake_videos = list(Path("/content/drive/MyDrive/data/face++dataset/ffpp_fake").glob("*.mp4"))
#Shuffling the fake and training datasets
random.shuffle(real_videos)
random.shuffle(fake_videos)
```

```python
train_size_real = int(len(real_videos) * 0.8)
train_size_fake = int(len(fake_videos) * 0.8)
train_videos_real = real_videos[:train_size_real]
train_videos_fake = fake_videos[:train_size_fake]
test_videos_real = real_videos[train_size_real:]
test_videos_fake = fake_videos[train_size_fake:]
train_split_real = int(len(train_videos_real) * 0.8)
train_split_fake = int(len(train_videos_fake) * 0.8)
final_train_videos_real = train_videos_real[:train_split_real]
final_train_videos_fake = train_videos_fake[:train_split_fake]
val_videos_real = train_videos_real[train_split_real:]
val_videos_fake = train_videos_fake[train_split_fake:]
```

### **Extract frames from video**

```python
def extract(path_video_list, Route, dataset_name):

    os.makedirs(Route, exist_ok=True)

    for path_video in tqdm(path_video_list,desc=f"Extracting {dataset_name}",unit="video"):

        cap = cv.VideoCapture(str(path_video))

        video_name = Path(path_video).stem

        total_frames = int(cap.get(cv.CAP_PROP_FRAME_COUNT))

        if total_frames <= 0:
            cap.release()
            continue

        frame_indices = np.linspace(0,total_frames - 1,5,dtype=int)

        frames_extracted = 0

        for i, frame_idx in enumerate(frame_indices):

            cap.set(cv.CAP_PROP_POS_FRAMES, frame_idx)

            ret, frame = cap.read()

            if ret:

                save_path = f"{Route}/{video_name}_{i}.jpg"

                cv.imwrite(save_path, frame)

                frames_extracted += 1

        cap.release()
```

```python
Router_path1_train = "/content/drive/MyDrive/frame_train/frame_real"
Router_path2_train = "/content/drive/MyDrive/frame_train/frame_fake"

Router_path1_val = "/content/drive/MyDrive/frame_validation/frame_real"
Router_path2_val = "/content/drive/MyDrive/frame_validation/frame_fake"

Router_path1_test = "/content/drive/MyDrive/frame_test/frame_real"
Router_path2_test = "/content/drive/MyDrive/frame_test/frame_fake"

extract(final_train_videos_real,Router_path1_train,"Train - Real")
extract(final_train_videos_fake,Router_path2_train,"Train - Fake")

extract(val_videos_real,Router_path1_val,"Validation - Real")
extract(val_videos_fake,Router_path2_val,"Validation - Fake")

extract(test_videos_real,Router_path1_test,"Test - Real")
extract(test_videos_fake,Router_path2_test,"Test - Fake")
```
